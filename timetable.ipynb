```python
# 13.1
import time
fmt = "%Y-%m-%d"
t = time.localtime()


today = open("today.txt", "w")
today.write(time.strftime(fmt, t))
today.close()
```

```python
# 13.2
today = open("today.txt", "r")
today_string = today.read()
today.close()
```

```python
# 13.3
time.strptime(today_string, fmt)
```

```python
# 15.1
# Code only works in the multiprocess.py file.
import multiprocessing
import random
from datetime import datetime
import time


def whatTimeIsIt(num):
    randSec = random.random()
    time.sleep(randSec)
    print(f"I'm process {num}, and It's {datetime.now()}.")

if __name__ == "__main__":
    for n in range(3):
        p = multiprocessing.Process(target=whatTimeIsIt, args=str(n))
        p.start() 

```
